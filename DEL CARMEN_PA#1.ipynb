```python
def rotate_word(text): #Defining function
    return text[1:] + text[0] #Code to put first index to the end 

print(f"programming -> {rotate_word('programming')}") #Example to show that the code works
print(f"Ece -> {rotate_word('Ece')}")
print(f"Code -> {rotate_word('Code')}")
print(f"Cat -> {rotate_word('A')}")
print(f"Clark -> {rotate_word('Clark')}")
print()
```

```python
def make_username(first_name, last_name): #Defining the function
    firstname_changed = first_name.lower().replace(" ", "") #Code to turn the first name to lower case
    lastname_changed = last_name.lower().replace(" ", "") #Code to turn the last name to lower case and replace space with dot
    return firstname_changed + "." + lastname_changed

print(f"Clark, Del Carmen -> {make_username('Clark', 'Del Carmen')}") #Examples to show that the code works
print(f"LeBron, James -> {make_username('LeBron', 'James')}")
print(f"Juan, 'Dela Cruz -> {make_username('Juan', 'Dela Cruz')}")
print()
```

```python
def swap_bookends(items): #Defining the function 
    first, *middle, last = items #Code t
    return [last] + middle + [first]

print(f"7,8,9,10 -> {swap_bookends([7,8,9,10])}")
print(f"yellow, orange, black -> {swap_bookends(['yellow', 'orange', 'black'])}")
print(f"24, 8 -> {swap_bookends([24, 8])}")
print()
```
